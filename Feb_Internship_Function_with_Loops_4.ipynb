Problem 1: Mobile Recharge Validation System
 Problem Statement
Create a function to validate a mobile recharge request.
Rules:
Recharge amount must be ≥ ₹50


Valid plans: 199, 299, 399, 599


User should be allowed to retry if invalid


Real-Time Use
Telecom recharge apps


Payment gateways


Hint
Use function for validation


Use while loop for retry


```python
def mobile_recharge_validation(recharge_amount):
    valid_plans = [199, 299, 399, 599]
    if recharge_amount <= 50:#checking if recharge amount is less than 50
      print("Recharge amount should be greater than or equal to ₹50.")
      return False
    elif recharge_amount not in valid_plans:
      print("Invalid plan selected. Please select a valid plan.")
      return False
    else:
      print("Recharge is successful")
      return True
while(True):
  recharge_amount=int(input("Enter the recharge amount:"))#taking input
  if mobile_recharge_validation(recharge_amount):
    break
  else:
    print("Invalid recharge amount. Please try again.")
```

Problem 2: Inventory Reorder Alert System
Problem Statement
Given a dictionary of products and stock quantities:
Print Reorder Alert if stock < 15


Print Stock OK otherwise


Real-Time Use
Warehouse systems


Retail management


Hint
Use function


Use for loop on dictionary


```python
def inventory_reorder_alert(inventory):
  for product, quantity in inventory.items():
    if quantity < 15:#checking if quatity is more than 15
      print(f"Reorder alert: {product} (Quantity: {quantity})")#if less then printing reorder
    else:
      print(f"{product} (Quantity: {quantity}) - Stock OK")#if >=15 printing stock ok

inventory = {"Sugar":25,"Salt":10,"Oil":7,"Dal":16}
inventory_reorder_alert(inventory)
```

 Problem 3: Student Result Processing System
Problem Statement
Create a function that:
Accepts a list of student marks


Calculates average marks


Prints Pass if average ≥ 50 else Fail


Real-Time Use
Online exam systems


Evaluation portals


Hint
Use loop to calculate total


Return result from function




```python
def result_processing_system(marks):
  total = 0#intialize total value
  for mark in marks:
    total = total + mark#calculating total
  average = total / len(marks)#calculating average
  if(average >= 50):#checking if >=50 and printing the stmts accordingly
    return "Pass"
  else:
    return "Fail"
marks=[30,40,70,60,20]
result = result_processing_system(marks)
print("Result:",result)#result is printed
```

Problem 4: Cab Fare Estimator with Retry Option
Problem Statement
Create a cab fare calculator:
Base fare = ₹50


₹12 per km


Peak hour → 25% extra


Allow retry using while loop


Real-Time Use
Ride-hailing apps


Transport systems


Hint
Function for fare calculation


While loop for retry




```python
def cab_fare_calculator():
  while True:
    #taking input
    distance = int(input("Enter the distance in kilometers: "))
    peak_hour = input("Is it a peak hour? (yes/no): ")
    fare=50+(distance*12)#calculating fare
    if peak_hour.lower() == "yes":#calculating fare in peak time
      fare = fare * 1.25
      print("Total fare:",fare)
    else:
        print("Total fare:",fare)
    retry=input("Do you want to retry? (yes/no): ")#checkin if want to retry
    if retry.lower() == "no":
        break
cab_fare_calculator()
```

Problem 5: Employee Attendance Eligibility Checker
Problem Statement
Create a function that:
Accepts attendance list ("P", "A")


Calculates attendance percentage


Returns Eligible if ≥ 75% else Not Eligible


Real-Time Use
HR payroll systems


Internship tracking


 Hint
Use loop to count present days


Use return statement




```python
def attendence_checker(attendence):
  p = 0#intializing present count to 0
  for i in attendence:
    if i == "P":
      p = p + 1#incrementing
  attendance_percentage = p / len(attendence) * 100#calculating percentage
  if attendance_percentage >= 75:#checking percentage and printing relevant stmts
    return "Eligible"
  else:
    return "Not Eligible"
attendence = ["P", "P", "A", "P", "A", "P", "A", "P", "A", "P"]
result = attendence_checker(attendence)
print(result)
```

Problem 6: Password Strength Checker
Problem Statement
Create a function to check password strength.
Rules:
Minimum length = 8


Must contain at least one digit


Must contain at least one special character (@#$)


Real-Time Use
Login systems


Security validation


Hint
Use loop over string


Use conditions inside function


```python
def check_password(p):
  #intializing count of digit and special charaters in input
  digit = 0
  special = 0
  for i in p:
   if i >= '0' and i <= '9':#checking input for digits
    digit = 1
   if i == '@' or i == '#' or i == '$':#checking input for speacial characters
    special = 1
  if len(p) >= 8 and digit == 1 and special == 1:#checking length and all constraints
        print("Strong Password")
  else:
        print("Weak Password")
password = input("Enter password: ")
check_password(password)
```
