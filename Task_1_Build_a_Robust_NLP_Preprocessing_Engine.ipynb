Task 2:Build Advanced Preprocessing Function
Implement the following function:
def preprocess_text(text):
    pass

Required Features
Your function must include:
Remove numbers
Example: "I have 2 dogs" → "I have dogs"
Remove extra spaces
Example: "This is good" → "This is good"
Handle repeated characters
Example: "soooo goooood!!!" → "so good"
Convert text to lowercase
Example: "WOW!!! This is GREAT!!!" → "wow this is great"
Remove very short tokens
Remove words with length ≤ 2
Exception: keep meaningful words like "no", "not"
Remove URLs and email-like patterns
Example: "Visit http://example.com now" → "visit now"


```python
import re

def preprocess_text(text):
    if not text or text.isspace():
        return [], ""

    # Convert to lowercase
    text = text.lower()

    # Remove URLs and emails
    text = re.sub(r'\S+@\S+|\bhttps?://\S+\b', '', text)

    # Remove numbers
    text = re.sub(r'\d+', '', text)

    # Replace repeated characters (more than 2) with 2
    text = re.sub(r'(.)\1{2,}', r'\1\1', text)

    # Remove extra spaces and punctuation (except meaningful ones)
    text = re.sub(r'[^\w\s]', '', text)
    text = re.sub(r'\s+', ' ', text).strip()

    # Tokenize
    tokens = text.split()

    # Remove very short tokens (length <=2) except meaningful ones
    meaningful = {'no', 'not'}
    tokens = [t for t in tokens if len(t) > 2 or t in meaningful]

    cleaned_sentence = ' '.join(tokens)
    return tokens, cleaned_sentence
```

Task 4: Token Analytics
For each processed sentence, compute:
Total number of tokens
Number of unique tokens
Average token length
Analysis Questions
Which sentence had the most noise?
Which sentence retained the most meaningful tokens after cleaning?


```python
for text in sample_inputs:
    tokens, clean_sentence = preprocess_text(text)
    total_tokens = len(tokens)
    unique_tokens = len(set(tokens))
    avg_token_length = round(sum(len(t) for t in tokens)/total_tokens, 2) if total_tokens else 0
    print(f"Sentence: {clean_sentence}")
    print(f"Total Tokens: {total_tokens}, Unique Tokens: {unique_tokens}, Avg Token Length: {avg_token_length}")
    print("-"*60)
```

Task 5: Frequency Analysis
Combine all tokens and use:
from collections import Counter

Identify:

```python
from collections import Counter

all_tokens = []
for text in sample_inputs:
    tokens, _ = preprocess_text(text)
    all_tokens.extend(tokens)

counter = Counter(all_tokens)
print("Top 10 most frequent words:", counter.most_common(10))
print("Top 5 least frequent words:", counter.most_common()[-5:])
```

Task 6: Build Full Pipeline
def full_pipeline(text_list):
    pass

Expected Output
{
    "tokens": [...],
    "clean_sentences": [...]
}



```python
def full_pipeline(text_list):
    all_tokens = []
    clean_sentences = []

    for text in text_list:
        tokens, clean_sentence = preprocess_text(text)
        all_tokens.extend(tokens)
        clean_sentences.append(clean_sentence)

    return {"tokens": all_tokens, "clean_sentences": clean_sentences}

pipeline_output = full_pipeline(sample_inputs)
pipeline_output
```
