```python
from langchain_core.prompts import PromptTemplate, ChatPromptTemplate
```

TASK 1: Replace Hardcoded Prompt

```python
explain_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} in simple terms for beginners"
)

def explain_topic(topic):
    return explain_template.format(topic=topic)

print(explain_topic("Artificial Intelligence"))
```

TASK 2: Multi-Input Prompt Syste

```python
multi_input_template = PromptTemplate(
    input_variables=["topic", "audience", "tone"],
    template="Explain {topic} for {audience} in a {tone} tone"
)

def generate_multi_prompt(topic, audience, tone):
    return multi_input_template.format(
        topic=topic,
        audience=audience,
        tone=tone
    )

print(generate_multi_prompt("AI", "beginners", "friendly"))
print(generate_multi_prompt("Python", "kids", "fun"))
print(generate_multi_prompt("Deep Learning", "engineers", "technical"))
```

TASK 3: Prompt Variations Engine

```python
teaching_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} clearly step by step"
)

interview_template = PromptTemplate(
    input_variables=["topic"],
    template="Ask 3 interview questions about {topic}"
)

storytelling_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} as a story"
)

def run_variations(topic):
    return {
        "teaching": teaching_template.format(topic=topic),
        "interview": interview_template.format(topic=topic),
        "storytelling": storytelling_template.format(topic=topic)
    }

print(run_variations("Machine Learning"))
```

TASK 4: ChatPromptTemplate System

```python
chat_template = ChatPromptTemplate.from_messages([
    ("system", "You are a {role} who responds based on your role."),
    ("user", "Explain {topic}")
])

def chat_prompt(role, topic):
    return chat_template.format_messages(role=role, topic=topic)

print(chat_prompt("teacher", "Neural Networks"))
print(chat_prompt("interviewer", "Neural Networks"))
print(chat_prompt("motivator", "Neural Networks"))
```

TASK 5: Input Validation Layer

```python
def validate_inputs(audience, tone):
    valid_audience = ["beginner", "intermediate", "expert"]
    valid_tone = ["formal", "casual", "fun"]

    if audience not in valid_audience:
        audience = "beginner"

    if tone not in valid_tone:
        tone = "casual"

    return audience, tone

print(validate_inputs("pro", "excited"))
```

TASK 6: Prompt Generator App

```python
def generate_prompt(topic, audience, tone, style):

    audience, tone = validate_inputs(audience, tone)

    teaching = PromptTemplate(
        input_variables=["topic", "audience", "tone"],
        template="Explain {topic} for {audience} in a {tone} tone in a teaching style"
    )

    interview = PromptTemplate(
        input_variables=["topic"],
        template="Create interview questions about {topic}"
    )

    storytelling = PromptTemplate(
        input_variables=["topic", "audience"],
        template="Explain {topic} as a story for {audience}"
    )

    if style == "teaching":
        return teaching.format(topic=topic, audience=audience, tone=tone)

    elif style == "interview":
        return interview.format(topic=topic)

    elif style == "storytelling":
        return storytelling.format(topic=topic, audience=audience)

    else:
        return teaching.format(topic=topic, audience=audience, tone=tone)

print(generate_prompt("Neural Networks", "beginner", "fun", "storytelling"))
```

TASK 7: Template Reusability Test

```python
reusable_template = PromptTemplate(
    input_variables=["topic", "audience", "tone"],
    template="Explain {topic} for {audience} in a {tone} tone"
)

test_cases = [
    ("AI", "beginner", "fun"),
    ("AI", "intermediate", "formal"),
    ("AI", "expert", "casual"),
    ("AI", "beginner", "casual"),
    ("AI", "expert", "fun")
]

for topic, audience, tone in test_cases:
    print(reusable_template.format(
        topic=topic,
        audience=audience,
        tone=tone
    ))
```
