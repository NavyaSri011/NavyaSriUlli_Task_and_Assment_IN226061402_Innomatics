Problem 1: Smart Light Controller
Problem Statement
Create a class to control a smart light that can be turned ON or OFF and display its status.
🌍 Real-Time Use
Smart home automation systems
📥 Sample Input
Light Name: Bedroom Light
Action: ON
📤 Expected Output
Bedroom Light is ON



```python
class SmartLightController:#defining the class
    def __init__(self, light_name):#constructor
        self.light_name = light_name
        self.status = "OFF"
    def turn_on(self):#method for ON
        self.status = "ON"
    def turn_off(self):#Method for OFF
        self.status = "OFF"
    def display_status(self):
        print(f"{self.light_name} is {self.status}")
light_name=input("Light Name: ")#Taking input from user
action=input("Action: ")
smart_light=SmartLightController(light_name)
#Checking the conditions accoring to the user input
if action=="ON":
    smart_light.turn_on()
elif action=="OFF":
    smart_light.turn_off
smart_light.display_status()
```

Problem 2: Employee ID Card System
 Problem Statement
Create a class that stores employee name, ID, and department and displays ID card details.
Real-Time Use
Corporate HR systems
📥 Sample Input
Employee Name: Rahul
Employee ID: EMP102
Department: IT

📤 Expected Output
Employee ID Card
Name: Rahul
ID: EMP102
Department: IT



```python
class EmployeeIDCard:#defining a class
    def __init__(self, name, id, department):#constuctor
        self.name = name
        self.id = id
        self.department = department
    def display_id_card(self):#creating method
        print("Employee ID Card")
        print(f"Name: {self.name}")
        print(f"ID: {self.id}")
        print(f"Department: {self.department}")
        #taking inputs
name=input("Employee Name: ")
id=input("Employee ID: ")
department=input("Department: ")
employee_id_card=EmployeeIDCard(name, id, department)#creating object
employee_id_card.display_id_card()#calling
```

Problem 3: Mobile Contact Record
Problem Statement
Create a class to store contact name and phone number and display contact information.
Real-Time Use
Mobile contact applications
📥 Sample Input
Contact Name: Anita
Phone Number: 9876543210
📤 Expected Output
Contact Saved
Name: Anita
Phone: 9876543210



```python
class ContactRecord:#creating a class
  def __init__(self, name, phone_number):#constructor
    self.name = name
    self.phone_number = phone_number
  def display_contact_info(self):#creating a method
    print("Contact Saved")
    print(f"Name: {self.name}")
    print(f"Phone: {self.phone_number}")
    #taking input from user
name=input("Contact Name: ")
phone_number=input("Phone Number: ")
contact_record=ContactRecord(name, phone_number)#creating an object
contact_record.display_contact_info()#calling
```

Problem 4: Product Price Tag Generator
Problem Statement
Create a class that stores product name and price and prints a formatted price tag.
 Real-Time Use
Retail billing systems
📥 Sample Input
Product Name: Headphones
Price: 2499
📤 Expected Output
Product: Headphones
Price: ₹2499


```python
class ProductPriceTagGenerator:
  def __init__(self,product_name,price):
    self.product_name=product_name
    self.price = price  # Corrected: assign price to self.price
  def display_record(self):
    print(f"Product: {self.product_name}")
    print(f"Price: ₹{self.price}")
product=input("Product Name: ") # Corrected: input for product name
price=int(input("Price:"))
product_price_tag=ProductPriceTagGenerator(product,price)
product_price_tag.display_record()
```

Problem 5: Movie Rating Display System
Problem Statement
Create a class that stores movie name and rating and displays movie details.
Real-Time Use
Streaming platforms
📥 Sample Input
Movie Name: Inception
Rating: 4.8
📤 Expected Output
Movie: Inception
Rating: 4.8 / 5


```python
class MovieRatingDisplaySystem:#creating class
  def __init__(self,movie_name,rating):#defining constructor
    self.movie_name=movie_name
    self.rating=rating
  def display_record(self):#defining method
    print(f"Movie: {self.movie_name}")
    print(f"Rating: {self.rating} / 5")
    #taking input
movie=input("Movie Name: ")
rating=float(input("Rating: "))
movie_rating_genera=MovieRatingDisplaySystem(movie,rating)#creating object
movie_rating_genera.display_record()#calling object
```
