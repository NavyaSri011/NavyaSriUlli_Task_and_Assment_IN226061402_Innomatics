Problem Statement 1:Smart Parking Lot Management System
Design a function to manage a smart parking lot.
The system should:
Accept vehicle entry and exit logs


Calculate total parked vehicles


Identify peak parking usage


Alert if parking exceeds capacity


 Real-Time Use
Mall parking systems


Smart city infrastructure


 Hint
Function + list


Loop for counting


Conditional alert
Sample Input
Parking Capacity: 50
Vehicle Logs: ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"]

Expected Output
Currently Parked Vehicles: 3
Parking Status: Available
  

```python
def parking_lot_management():#defining function
  #taking input
  n=int(input("Parking Capacity:"))
  x=input("Vehicle log:")
  y=x.split()
  ct=0
  count=0
  for action in y:
    if action =="IN":#checking if in or out
      ct=ct+1   #and incrementing accordingly
    elif action=="OUT":
      count=count+1
  v=ct-count
  u=n-v#calculating if there is availble space to park
  print("Currently Parked Vehicles:",v)
  if u>0:
    print("Parking Status:Available")
  else:
    print("Parking Status:Not Available")
parking_lot_management()
```

50%–80% → Warning


80% → Critical



 Real-Time Use

DevOps alerting systems


 Hint
Function + loop


Average calculation

CPU Readings (%): [45, 60, 70, 85, 90]

Expected Output

Server Status: Warning




```python
def cloud_server_load_classification_system():#defining function
  n=list(map(int,input("CPU Readings(%):").split()))#taking list as input
  average=sum(n)/len(n)#calculating avg
  print("Average CPU Load:",average,"%")
  if average<50:#checking the conditions and printing the stmts accordingly
      print("Server Status:Normal")
  elif 50<=average<=80:
    print("Server Status:Warning")
  elif average>80:
    print("Server Status:Critical")
cloud_server_load_classification_system()#calling the function
```

 Problem Statement 5:Smart Classroom Resource Usage Monitor
Design a function that tracks usage of classroom resources (projector, AC, lights) and identifies overuse patterns.
Real-Time Use
Smart classrooms


Energy optimization systems


Hint
Function + dictionary


Conditional checks

Sample Input
Resource Usage (hours):
{
 "Projector": 6,
 "AC": 9,
 "Lights": 4
}

Expected Output
Overused Resources: AC
Energy Alert: Yes



```python
def resource_usage_monitor():#defining function
  usage={}#considering a dictionary
  print("Reource Usage(hours):")
  usage["Projector"]=int(input("projector:"))
  usage["AC"]=int(input("AC:"))
  usage["Lights"]=int(input("Lights:"))
  overused=[]#defining a list for overused items to store
  threshold=7   #taking threshold as 7
  for resource,hours in usage.items():
    if hours>threshold:#checking conditions
      overused.append(resource)
  if overused:
      print("Overused Resources:",",".join(overused))
      print("Engery Alert:Yes")
  else:
      print("Overused Resources:None")
      print("Engery Alert:No")
resource_usage_monitor()#calling the function
```
