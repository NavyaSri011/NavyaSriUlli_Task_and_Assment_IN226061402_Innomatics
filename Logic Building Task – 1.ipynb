2. Pass / Fail Analyzer
Problem Statement:
Analyze student results.
marks = [45, 78, 90, 33, 60]

Requirements:
A student passes if marks ≥ 50
Count the total number of pass students
Count the total number of fail students
Print the final result clearly
Real-World Application: Academic evaluation systems

```python
#Given students marks array
marks=[45,78,90,33,60]
#Initializing the count for pass and fail
pass_students_count=0
fail_students_count=0
for i in marks:
  if i>=50:
    pass_students_count=pass_students_count+1 #incrementing the count if pass
  else:
    fail_students_count=fail_students_count+1 #incrmenting count if fail
    #printing final results
print("Total number of pass students:",pass_students_count)
print("Total number of fail students:",fail_students_count)
```

3. Simple Data Cleaner
Problem Statement:
Clean and standardize user names.
names = [" Alice ", "bob", " CHARLIE "]

Requirements:
Remove extra spaces
Convert all names to lowercase
Print the cleaned list
Real-World Application: Data preprocessing before analysis


```python
#Given names array
names = ["Alice ","bob","CHARLIE"]
for i in range(len(names)):
  names[i]=names[i].strip() #removing extra spaces
  names[i]=names[i].lower() #converting to lowercase
print(names) #printing the cleaned list
```


4. Message Length Analyzer
Problem Statement:
Analyze message sizes.
messages = ["Hi", "Welcome to the platform", "OK"]

Requirements:
Print the length of each message
Flag messages longer than 10 characters
Real-World Application: Text filtering and validation systems


```python
#Given
messages=["Hi", "Welcome to the platform", "OK"]
#printing length
for message in messages:
  print("Your message is:", message)
  print("Length of message is:",len(message))
  if(len(message)>10):#checking if charcter is longer than 10 characters
    print("Flagged: Message is longer than 10 chacters")
```

5. Error Message Detector
Problem Statement:
Detect error messages from system logs.
logs = ["INFO", "ERROR", "WARNING", "ERROR"]

Requirements:
Count the number of "ERROR" entries
Print the total error count
Real-World Application: Monitoring and log analysis systems


```python
#Given
logs=["INFO", "ERROR", "WARNING", "ERROR"]
#initializing count
error_count=0
for log in logs:
  if(log=="ERROR"):#checking list for error
    error_count=error_count+1 #incrementing error count
print("Total error count is:", error_count)
```
