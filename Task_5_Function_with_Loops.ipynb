Problem 1: Social Media – Post Engagement Analyzer
Problem Statement
Create a function to analyze engagement on a social media post.
Rules
Calculate total likes
If Total Likes ≥ 1000 → Viral Post
Otherwise → Normal Engagement
Real-Time Use
Social media analytics platforms
Marketing dashboards
Hint
Use function
Use loop for calculations
Expected Output Format
Total Likes: 1050
Post Status: Viral Post




```python
def social_media(likes):#defining function
  total_likes = 0
  for like in likes:
    total_likes = total_likes + like#calculating total likes
  print("Total Likes:", total_likes)
  if total_likes >= 1000:#checking condition
    print("Post Status: Viral Post")
  else:
    print("Post Status: Normal Engagement")
social_media([500,300,250])#passing values to the function
```

Problem 3: Agriculture – Rainfall Adequacy Checker
Problem Statement
Create a function that analyzes seasonal rainfall data.
Rules
Calculate average rainfall
Average ≥ required level → Adequate Rainfall
Otherwise → Inadequate Rainfall
Real-Time Use
Climate monitoring systems
Irrigation planning tools
Hint
Use function
Use loop and math operations
Expected Output Format
Average Rainfall: 72
Rainfall Status: Adequate Rainfall


```python
def rainfall_checker(rainfall,required_level):#defining function
  total_rainfall = 0
  for rain in rainfall:
    total_rainfall = total_rainfall + rain#calculating total rainfall
  average_rainfall = total_rainfall / len(rainfall)#calculating average rainfall
  print("Average Rainfall:", average_rainfall)
  if average_rainfall >= required_level:#checking condition
    print("Rainfall Status: Adequate Rainfall")
  else:
    print("Rainfall Status: Inadequate Rainfall")
rainfall_checker([10,20,30,40,50],50)#passing values to the function
```

Problem 4: Social Media – Duplicate Account Detection
Problem Statement
Create a function to detect duplicate usernames in a system.
Rules
If duplicates exist → Flag accounts
Else → Accounts are unique
Real-Time Use
Platform moderation
Fraud detection systems
Hint
Use function
Use set data structure
Expected Output Format
Duplicate Accounts Found: Yes



```python
def duplicate_account(users):#function declartion
  usernames = set()#taking an empty set
  flag=False#considering flag
  for user in users:
    if user in usernames:
      flag=True
      break
    else:
      usernames.add(user)#comparing empty set with users in order to findout duplicate values
  if flag:
    print("Duplicate Accounts Found: Yes")
  else:
    print("Duplicate Accounts Found: No")
duplicate_account(["yash123","evva","lisa","yash123"])#passing values to the function
```

Problem 6: Agriculture – Premium Crop Price Filter

Create a function that filters crop prices above a premium threshold.

Else → Regular Crop
Real-Time Use

Hint
Use function
Use list filtering logic

```python
def price_filter(crop_prices):#defining function
  premium_crops = []#empty list
  for crop_price in crop_prices:
    if crop_price > 2000:#checking base condition
      premium_crops.append(crop_price)
  print("Premium Crops:", premium_crops)
price_filter([1500,1050,2500,3200,1986])#passing value to function
```

Problem 9: E-Learning – Student Attendance Eligibility System
Problem Statement
Create a function to calculate attendance percentage and check exam eligibility.
Rules
Attendance ≥ 75% → Eligible
Attendance < 75% → Not Eligible
Real-Time Use
Learning management systems
Internship tracking platforms
Hint
Use function
Use list and loop
Expected Output Format
Attendance Percentage: 80.0
Exam Eligibility: Eligible



```python
def attendence_eligibilty(attendence):#defining function
  total_present_days = 0
  for att in attendence:
    if att == 'P':
      total_present_days += 1 # Incrementing count
  if not attendence:
    attendence_percentage = 0
  else:
    attendence_percentage = (total_present_days/len(attendence))*100#calculating attendence percentage
  print("Attendance Percentage:", attendence_percentage)
  if attendence_percentage >= 75:#checking condition
    print("Exam Eligibility: Eligible")
  else:
    print("Exam Eligibility: Not Eligible")
attendence_eligibilty(['P','P','P','P','A'])
```

Problem 10: Smart Electricity Bill Analyzer
Problem Statement
You are building a smart electricity billing system for residential users.
Create a function that calculates the electricity bill and classifies usage level based on units consumed.
Billing Rules
First 100 units → ₹3 per unit
Next 100 units (101–200) → ₹5 per unit
Above 200 units → ₹7 per unit
Usage Classification
Total bill < ₹500 → Low Usage
₹500 – ₹1500 → Moderate Usage
₹1500 → High Usage
Real-Time Use
Smart meter systems
Electricity board billing software
Energy consumption analytics
Hint
Use function to calculate bill
Use conditional logic for slabs
Use return values (bill and usage status)


```python
def electricity_bill_analyzer(units):
  bill = 0
  if units <= 100:#100 units 3rs per unit
    bill = units * 3
  elif units <= 200: #(101-200)units 5rs per unit
    bill = units*5
  else: # Above 200 units at 7rs per unit
    bill = units*7
  print("Total Bill:", bill)
  if bill < 500:
    print("Usage Status: Low Usage")
  elif 500 <= bill < 1500:
    print("Usage Status: Moderate Usage")
  else:
    print("Usage Status: High Usage")
electricity_bill_analyzer(400)#passing values
```
