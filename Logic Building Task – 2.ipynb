Problem 1: Unique Words in a Sentence
Problem Statement
Given a sentence, identify all unique words and display:
The count of unique words
The unique words themselves
Hint
Split the sentence into individual words.
Convert the list of words into a set to remove duplicates.
Print both the set and its length.
Input
Enter sentence: python is easy and python is powerful

Expected Output
Unique words count: 5  
Unique words: {'python', 'is', 'easy', 'and', 'powerful'}

Real-Time Applications
Text analytics
Natural Language Processing (NLP) preprocessing
Search engine indexing
Chat analysis systems


```python
sentence=input("Enter sentence:")#Taking input from user
words = sentence.split()
unique_words=set(words)#converting list to set
print("Unique words count: ",len(unique_words))
print("Unique words : ",unique_words)
```

Problem 2: Highest Salary from Employee Data
Problem Statement
Store employee names and their salaries, and determine which employee has the highest salary.
Hint
Use a dictionary where the key is the employee name and the value is the salary.
Compare salary values to find the highest.
Input Format
A dictionary where:
Keys represent employee names
Values represent employee salaries
Example Input
employees = {
    "Ravi": 75000,
    "Anita": 68000,
    "Kiran": 72000
}

Expected Output
Highest Salary: Ravi - 75000

Real-Time Applications
HR management systems
Payroll analysis


```python
n=int(input("Enter the number of employees:"))#Taking input
employees_data={}#initializing dict
for i in range(n):
  employees=input("Enter employee name: ")
  employees_sal=int(input("Enter employee salary: "))
  employees_data[employees]=employees_sal
highest_emp=max(employees_data, key=employees_data.get)
#getting the highest salary
highest_sal=employees_data[highest_emp]
print("Highest Salary:",highest_emp, "-", highest_sal)
```

Problem 3: Find Maximum and Minimum Values
Problem Statement
Find the maximum and minimum values from a list of numbers.
Hint
Initialize both maximum and minimum with the first element of the list.
Use a loop to compare each element.
Input Format
A list of integers where each element represents a numeric value.
Example Input
numbers = [45, 22, 89, 10, 66]

Expected Output
List: [45, 22, 89, 10, 66]  
Max: 89  
Min: 10

Real-Time Applications
Sales data analysis
Temperature monitoring systems
Performance metric evaluation



```python
#getting input
n = int(input("Enter the number of elements:"))
lst=[]
for i in range(n):
  elements=int(input("Enter the elements:"))
  lst.append(elements)
max_element=lst[0]#initializing min anx max to 0
min_element=lst[0]
for i in range(n):
  if lst[i]>max_element:#comparing all the elements in the list
    max_element=lst[i] #returing max value
  elif lst[i]<min_element:#min value
    min_element=lst[i]
print("List:", lst)
print("Max:", max_element)
print("Min:", min_element)
```

Problem 4: Count Products Above a Price Threshold
Problem Statement
Count how many products have a price greater than 1000.
Hint
Use a loop to compare each price with 1000.
Input Format
A list of integers where each element represents the price of a product.
Example Input
prices = [450, 1200, 899, 1500, 300]

Expected Output
Products above 1000: 2

Real-Time Applications
E-commerce product filtering
Budget analysis


```python
#Given input
Input_prices=[450,1200,899,1500,300]
ct=0 #intializing the count variable
for i in range(len(Input_prices)):
  if(Input_prices[i] > 1000):   #checking if the price is >1000
    ct=ct+1       #increment the count variable by 1 if condiatin satisfys
print("Products above 1000:",ct) #printing the count
```

Problem 5: Calculate Attendance Percentage
Problem Statement
Calculate the attendance percentage from a list containing "P" (Present) and "A" (Absent).
Hint
Count the number of "P" values.
Divide by the total number of days.
Example Input
attendance = ["P", "P", "A", "P", "P"]

Expected Output
Attendance Percentage: 80.0

Real-Time Applications
Internship tracking systems
HR attendance management systems


```python
#Given input list
Input_attendance=["P","P","A","P","P"]
ct=0  #initializing ct
for i in range(len(Input_attendance)): #for i lies in range
  if(Input_attendance[i]=="P"): #checking if i = p
    ct=ct+1  #incrementing ct
  per=(ct/len(Input_attendance))*100  #calculating percentage for no.of "P"
print("Attendance Percentage:",per)
```


Problem 7: Count Character Frequency
Problem Statement
Count how many times each character appears in a given string.
Hint
Use a dictionary.
Increment the count for each character occurrence.
Expected Output
{'p': 2, 'y': 1, 't': 1, 'h': 1, 'o': 1, 'n': 1}

Real-Time Applications
Text analysis
Password strength validation


```python
#taking input
input_str = input("Enter a string: ")
char_fre = {}
for char in input_str:
    if char in char_fre:
        char_fre[char] += 1#incrementing the char value if it is repeating in char_fre
    else:
        char_fre[char] = 1
print(char_fre)
```

Problem 10: Calculate Average Marks
Problem Statement
Calculate the average marks of a student from a list of marks.
Hint
Calculate the sum of all values.
Divide by the total number of marks.
Expected Output
Average Marks: 72.5

Real-Time Applications
Student performance evaluation
Result processing systems



```python
n=int(input("Enter the no.of subjects:"))
marks_lst=[]
for i in range(n):
  marks=int(input("Enter the marks:"))#Taking lst as input from user
  marks_lst.append(marks)
  avg=sum(marks_lst)/len(marks_lst)#calculating the avg of given input marks
print(f"Average Marks: {avg:.1f}")#printing avg
```
